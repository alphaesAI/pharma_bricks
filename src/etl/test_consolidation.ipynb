```python
# Databricks notebook source
# DBTITLE 1,Define Consolidation Test JSON Payload
import json

test_consolidation_payload = {
  "FileIds": [
    {
      "DataGroupTrackingID": "TRACK_MEMBER_02",
      "DataGroupMappingId": "834MEMBER",
      "FileId": "1111",
      "FileLayoutID": "834",
      "FileLayoutDescription": "Standard834",
      "CurrentContainer": "/Volumes/pharma_catalog/bronze",
      "CurrentFolderPath": "processed_data",
      "ConsolidatedMappingFilePath": "/Workspace/Users/logi@openhealthagents.org/pharma_bricks/src/etl/datalake-dev/JSON/Consolidation",
      "ConsolidatedMappingFileName": "ConsolidationMember712.json",
      "ConsolidatedLayerDataModelFilePath": "/Workspace/Users/logi@openhealthagents.org/pharma_bricks/src/etl/datalake-dev/JSON/Consolidation/DataModels",
      "ConsolidatedLayerDataModel": "MemberDataModel.json",
      "ConsolidatedFolderPath": "/Volumes/pharma_catalog/bronze/processed_data/consolidated_delta/834member"
    }
  ]
}


# Convert the Python dictionary into a single compressed JSON text string
json_string_input = json.dumps(test_consolidation_payload)

# COMMAND ----------
```

```python
# DBTITLE 1,Execute Target Loop Consolidation Pipeline
try:
    print("Starting Delta Consolidation pipeline test execution...")
    
    pipeline_result = dbutils.notebook.run(
        path="./databricks-dev/Notebooks/DatalakeProcessing/LoopConsolidationFiles",
        timeout_seconds=600,  # 10 minutes
        arguments={"ConsolidationJSON": json_string_input}
    )
    
    print("Pipeline returned. Parsing result...")
    result_json = json.loads(pipeline_result)  # This is a LIST of file results!
    
    # LoopConsolidationFiles returns a list of results (one per file)
    if isinstance(result_json, list):
        print(f"\n✓ Pipeline completed. Processed {len(result_json)} file(s)\n")
        
        # Check each file's status
        for idx, file_result in enumerate(result_json, 1):
            file_id = file_result.get('FileID', 'unknown')
            status = file_result.get('Status', 'UNKNOWN')
            count = file_result.get('ConsolidatedCount', '0')
            error = file_result.get('ErrorMessage', '')
            
            if status == "SUCCESS":
                print(f"  File {idx} (FileID={file_id}): ✓ SUCCESS - {count} rows")
            else:
                print(f"  File {idx} (FileID={file_id}): ✗ FAILED - {error}")
        
        # Overall success check
        all_success = all(r.get('Status') == 'SUCCESS' for r in result_json)
        if all_success:
            print("\n✓ All files processed successfully!")
        else:
            print("\n✗ Some files failed - see details above")
    else:
        # Handle unexpected dict response (shouldn't happen)
        print("✗ Unexpected response format (expected list)")
        print(json.dumps(result_json, indent=2))

except Exception as e:
    print("Pipeline execution encountered a critical error:")
    print(str(e))
```
